```python
import random, warnings, gc
from pathlib import Path
from collections import Counter
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader, WeightedRandomSampler
import torchvision
from sklearn.metrics import (
    roc_auc_score, average_precision_score, accuracy_score,
    precision_score, recall_score, f1_score, confusion_matrix,
    precision_recall_curve, roc_curve
)
from sklearn.model_selection import StratifiedGroupKFold
import rasterio

warnings.filterwarnings("ignore")
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("PyTorch:", torch.__version__, "| Device:", DEVICE)
if DEVICE.type == "cuda":
    print("GPU:", torch.cuda.get_device_name(0))

# ============================================================
# 1. CONFIGURATION & PATHS
# ============================================================
CONFIG = {
    # ---- paths ----
    "data_root": Path(r"/kaggle/input/datasets/shreyansdeshpande/200x200/FINALCNNDATA_200"),
    "csv_path": Path(r"/kaggle/input/datasets/shreyansdeshpande/200x200/FINALCNNDATA_200/_split_reports/all_patches_metadata.csv"),
    "ckpt_dir": Path(r"/kaggle/working/v6_ckpts"),

    # ---- data ----
    "n_channels": 11,
    "patch_size": 200,
    "nodata_value": -9999.0,
    "lidar_clip": 8.0,                      # z-score clamp, prevents outlier-driven overflow
    "channel_names": ["Blue", "Green", "Red", "NIR", "LRM", "Slope", "SVF", "HS1", "HS2", "HS3", "HS4"],
    "num_workers": 0,

    # ---- model ----
    "arch": "resnet50",                     # resnet18 | resnet34 | resnet50
    "ssl_arch": "resnet50",                 # architecture the SSL checkpoint was trained with
    "ssl_ckpt": Path(r"/kaggle/working/v6_ckpts/ssl_encoder.pt"),
    "bgr_to_rgb_init": True,                # data channels are B,G,R,... but ImageNet conv1 expects R,G,B
    "dropout": 0.4,

    # ---- optimisation ----
    "batch_size": 32,
    "epochs": 30,
    "lr": 3e-4,
    "encoder_lr_mult": 0.3,                 # encoder LR = lr * mult (1.0 = same LR for everything)
    "weight_decay": 5e-3,
    "grad_clip": 1.0,
    "label_smoothing": 0.05,                # train loss only (0 = off)
    "monitor": "ap",                        # metric for best epoch / early stop / LR schedule: "ap" | "auc"
    "early_stop_patience": 8,
    "sched_patience": 2,                    # must be < early_stop_patience so a reduced LR gets a chance

    # ---- sampling ----
    "use_weighted_sampler": True,
    "count_weight_cap": 3.0,
    "count_weight_alpha": 0.3,

    # ---- augmentation (all applied to the training set only; set to 0 to disable) ----
    "aug_noise_std": 0.05,                  # gaussian noise on z-scored channels
    "aug_contrast": 0.10,                   # global scale jitter in [1-x, 1+x]
    "aug_cutout_p": 0.5,
    "aug_cutout_min": 20,
    "aug_cutout_max": 60,
    "aug_chan_drop_p": 0.10,                # per-channel probability of zeroing the channel

    # ---- evaluation ----
    "use_tta": True,                        # 8-way flip/rot90 averaging for final OOF + test predictions
    "n_folds": 5,
    "seed": 42,
}

CONFIG["ckpt_dir"].mkdir(parents=True, exist_ok=True)
C, P = CONFIG["n_channels"], CONFIG["patch_size"]
THR_GRID = np.arange(0.05, 0.951, 0.01)


def set_seed(s):
    random.seed(s); np.random.seed(s); torch.manual_seed(s)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(s)

set_seed(CONFIG["seed"])

# ============================================================
# 2. METADATA LOADING (STRICT PARSING)
# ============================================================
def load_dataset_records():
    csv_file = CONFIG["csv_path"]
    if not csv_file.exists():
        raise FileNotFoundError(f"Could not find metadata CSV at {csv_file}")

    print(f"[+] Reading metadata from: {csv_file.resolve()}")
    df = pd.read_csv(csv_file)
    df.columns = df.columns.astype(str).str.strip()

    fn_col = "filename" if "filename" in df.columns else "patch_name"
    df["_filename"] = df[fn_col].astype(str).str.strip().apply(lambda p: Path(p).name)

    if not pd.api.types.is_numeric_dtype(df["label"]):
        df["target"] = (
            df["label"].astype(str).str.strip().str.lower()
            .map({"site": 1, "no_site": 0})
        )
        unmapped = df["target"].isna()
        if unmapped.any():
            bad = df.loc[unmapped, "label"].unique()
            raise ValueError(f"Unmapped label values found: {bad[:10]}")
        df["target"] = df["target"].astype(int)
    else:
        df["target"] = pd.to_numeric(df["label"], errors="coerce")
        if df["target"].isna().any():
            raise ValueError("Found non-numeric or missing values in numeric 'label'.")
        df["target"] = df["target"].astype(int)

    mon_col = [c for c in df.columns if c.lower().startswith("monumen")][0]
    df["monuments"] = pd.to_numeric(df[mon_col], errors="coerce").fillna(0).astype(int)

    if "group_id" in df.columns:
        df["group"] = df["group_id"].astype(str).str.strip()
    elif "block_x" in df.columns and "block_y" in df.columns:
        df["group"] = df["block_x"].astype(str).str.strip() + "_" + df["block_y"].astype(str).str.strip()
    else:
        raise KeyError("CSV missing 'group_id' or spatial coordinates.")

    trainval_recs, test_recs = [], []

    all_tifs = [f for f in CONFIG["data_root"].rglob("*.*") if f.suffix.lower() in (".tif", ".tiff")]
    name_counts = Counter(f.name for f in all_tifs)
    dupes = {n: cnt for n, cnt in name_counts.items() if cnt > 1}
    if dupes:
        raise ValueError(f"Duplicate filenames found under data_root — basename matching would be ambiguous: {list(dupes.items())[:5]}")
    disk_files = {f.name: f for f in all_tifs}

    missing_count = 0
    for _, row in df.iterrows():
        base_name = row["_filename"]
        if base_name not in disk_files:
            missing_count += 1
            continue

        record = {
            "path": str(disk_files[base_name]),
            "label": int(row["target"]),
            "count": int(row["monuments"]),
            "group": str(row["group"]),
        }
        split_name = str(row.get("split", "train")).lower().strip()
        if split_name == "test":
            test_recs.append(record)
        else:
            trainval_recs.append(record)

    if missing_count:
        print(f"[!] Warning: {missing_count} metadata rows had no matching file on disk.")

    print(f"[+] Loaded: {len(trainval_recs)} TrainVal patches | {len(test_recs)} Test patches")
    return trainval_recs, test_recs

trainval_recs, test_recs = load_dataset_records()

# ============================================================
# 3. NODATA-AWARE LOADER, STATS, DATASET
# ============================================================
def load_raw(path):
    with rasterio.open(path) as src:
        arr = src.read().astype(np.float32)
    invalid = (~np.isfinite(arr)) | (arr == CONFIG["nodata_value"])
    arr[invalid] = np.nan
    return arr, invalid


def compute_channel_stats(recs, max_samples=1000):
    idx = np.random.default_rng(CONFIG["seed"]).choice(len(recs), size=min(max_samples, len(recs)), replace=False)
    s = np.zeros(C); s2 = np.zeros(C); n = np.zeros(C)
    for j, i in enumerate(idx):
        if j % 100 == 0:
            print(f"  ...channel stats: {j}/{len(idx)} samples processed")
        arr, _ = load_raw(recs[i]["path"])
        for c in range(C):
            v = arr[c][np.isfinite(arr[c])]
            s[c] += v.sum(); s2[c] += (v ** 2).sum(); n[c] += v.size
    mean = s / np.maximum(n, 1)
    std = np.sqrt(np.maximum(s2 / np.maximum(n, 1) - mean ** 2, 1e-8))
    return mean.astype(np.float32), std.astype(np.float32)

print("\nComputing per-channel stats on TrainVal (nodata-masked)...")
CH_MEAN, CH_STD = compute_channel_stats(trainval_recs)
print(f"[+] Channel means: {np.round(CH_MEAN, 3)}")
print(f"[+] Channel stds : {np.round(CH_STD, 3)}")


class TerrainDataset(Dataset):
    def __init__(self, records, augment=False):
        self.records = records
        self.mean = CH_MEAN.reshape(C, 1, 1)
        self.std = CH_STD.reshape(C, 1, 1)
        self.augment = augment

    def __len__(self):
        return len(self.records)

    def _augment(self, x):
        """x: (C+1, H, W) float32, last channel is the validity mask (geometry only)."""
        # --- geometric: 8 dihedral orientations, applied to image + mask together ---
        if random.random() < 0.5: x = np.flip(x, 2)
        if random.random() < 0.5: x = np.flip(x, 1)
        k = random.randint(0, 3)
        if k: x = np.rot90(x, k, axes=(1, 2))
        x = np.ascontiguousarray(x, dtype=np.float32)

        # --- intensity / occlusion: image channels only (view into x, modified in place) ---
        img = x[:C]
        if CONFIG["aug_noise_std"] > 0:
            img += np.random.normal(0, CONFIG["aug_noise_std"], img.shape).astype(np.float32)
        if CONFIG["aug_contrast"] > 0:
            img *= np.float32(random.uniform(1 - CONFIG["aug_contrast"], 1 + CONFIG["aug_contrast"]))
        if CONFIG["aug_cutout_p"] > 0 and random.random() < CONFIG["aug_cutout_p"]:
            s = random.randint(CONFIG["aug_cutout_min"], CONFIG["aug_cutout_max"])
            i, j = random.randint(0, P - s), random.randint(0, P - s)
            img[:, i:i + s, j:j + s] = 0.0                      # 0 == channel mean after z-scoring
        if CONFIG["aug_chan_drop_p"] > 0:
            for c in range(C):
                if random.random() < CONFIG["aug_chan_drop_p"]:
                    img[c] = 0.0
        np.clip(img, -CONFIG["lidar_clip"], CONFIG["lidar_clip"], out=img)
        return x

    def __getitem__(self, i):
        r = self.records[i]
        arr, invalid = load_raw(r["path"])

        mean_fill = np.broadcast_to(self.mean, arr.shape)
        arr = np.where(np.isfinite(arr), arr, mean_fill).astype(np.float32)
        arr = (arr - self.mean) / self.std
        arr = np.clip(arr, -CONFIG["lidar_clip"], CONFIG["lidar_clip"])

        valid = (~invalid.any(0)).astype(np.float32)
        if self.augment:
            stacked = np.concatenate([arr, valid[None]], 0)
            stacked = self._augment(stacked)
            arr, valid = stacked[:C], stacked[C]

        return (torch.from_numpy(np.ascontiguousarray(arr)).float(),
                torch.tensor(r["label"], dtype=torch.float32),
                torch.from_numpy(np.ascontiguousarray(valid)).float())


def make_sample_weights(records):
    labels = np.array([r["label"] for r in records])
    counts = np.array([r["count"] for r in records])
    n_pos, n_neg = (labels == 1).sum(), (labels == 0).sum()
    base = np.where(labels == 1, n_neg / max(n_pos, 1), 1.0).astype(np.float64)
    cw = np.where(labels == 1,
                  np.minimum(1 + CONFIG["count_weight_alpha"] * np.log1p(counts), CONFIG["count_weight_cap"]),
                  1.0)
    return base * cw


def make_loader(records, train):
    ds = TerrainDataset(records, augment=train)
    nw = CONFIG["num_workers"]
    if train:
        if CONFIG["use_weighted_sampler"]:
            w = make_sample_weights(records)
            sampler = WeightedRandomSampler(torch.as_tensor(w, dtype=torch.double), len(w), True)
            return DataLoader(ds, batch_size=CONFIG["batch_size"], sampler=sampler,
                              num_workers=nw, drop_last=True)
        return DataLoader(ds, batch_size=CONFIG["batch_size"], shuffle=True,
                          num_workers=nw, drop_last=True)
    return DataLoader(ds, batch_size=CONFIG["batch_size"] * 2, shuffle=False, num_workers=nw)

# ============================================================
# 4. METRICS, TRAIN/EVAL EPOCH, TTA PREDICTION
# ============================================================
def compute_metrics(labs, probs, thr=0.5):
    """Threshold-free (AUC, AP) and thresholded (P/R/F1/Acc) metrics; safe for single-class inputs."""
    labs = np.asarray(labs).astype(int)
    two = len(np.unique(labs)) == 2
    pred = (probs >= thr).astype(int)
    tn, fp, fn, tp = confusion_matrix(labs, pred, labels=[0, 1]).ravel()
    return {
        "auc": roc_auc_score(labs, probs) if two else float("nan"),
        "ap": average_precision_score(labs, probs) if two else float("nan"),
        "f1": f1_score(labs, pred, zero_division=0),
        "prec": precision_score(labs, pred, zero_division=0),
        "rec": recall_score(labs, pred, zero_division=0),
        "acc": accuracy_score(labs, pred),
        "tn": int(tn), "fp": int(fp), "fn": int(fn), "tp": int(tp),
    }


def _sanitize(p):
    """Replace non-finite probabilities but report how many there were."""
    bad = int((~np.isfinite(p)).sum())
    return np.nan_to_num(p, nan=0.5, posinf=1.0, neginf=0.0), bad


def run_epoch(model, loader, criterion, optimizer=None, scaler=None, train=False):
    """One pass over `loader`. Returns a dict of metrics (+ probs/labs).

    Notes on interpretation:
      * Reported loss always uses the TRUE labels (label smoothing only affects the gradient).
      * Train metrics come from the (weighted) sampler + augmentation + train-mode dropout, so
        train AP / precision / F1 are inflated by class prevalence. Compare train AUC vs val AUC,
        and look at trends, not absolute train-vs-val gaps in prec/F1.
    """
    model.train() if train else model.eval()
    tot, n_seen, n_bad, n_skipped = 0.0, 0, 0, 0
    probs, labs = [], []
    eps = CONFIG["label_smoothing"] if train else 0.0

    for x, y, v in loader:
        x, y, v = x.to(DEVICE), y.to(DEVICE), v.to(DEVICE)
        y_fit = y * (1 - eps) + 0.5 * eps if eps > 0 else y
        if train:
            optimizer.zero_grad(set_to_none=True)
        with torch.set_grad_enabled(train):
            with torch.amp.autocast("cuda", enabled=(DEVICE.type == "cuda")):
                logits = model(x, v)
            loss = criterion(logits.float(), y_fit)
            if train:
                if not torch.isfinite(loss):
                    n_skipped += 1
                    continue
                scaler.scale(loss).backward()
                scaler.unscale_(optimizer)
                nn.utils.clip_grad_norm_(model.parameters(), CONFIG["grad_clip"])
                scaler.step(optimizer)
                scaler.update()

        with torch.no_grad():
            rep_loss = criterion(logits.detach().float(), y) if eps > 0 else loss.detach()
        tot += rep_loss.item() * x.size(0)
        n_seen += x.size(0)

        p, bad = _sanitize(torch.sigmoid(logits.detach().float()).cpu().numpy())
        n_bad += bad
        probs.append(p)
        labs.append(y.detach().cpu().numpy())

    if n_skipped:
        print(f"    [!] Skipped {n_skipped} batch(es) with non-finite loss")
    if n_bad:
        print(f"    [!] {n_bad} non-finite prediction(s) replaced — model may be diverging")

    probs = np.concatenate(probs) if probs else np.array([])
    labs = np.concatenate(labs) if labs else np.array([])
    out = compute_metrics(labs, probs) if len(labs) else {k: float("nan") for k in ["auc", "ap", "f1", "prec", "rec", "acc"]}
    out.update({"loss": tot / max(n_seen, 1), "probs": probs, "labs": labs})
    return out


@torch.no_grad()
def predict_probs(model, loader, tta=False):
    """Final predictions. With tta=True, averages the 8 flip/rot90 orientations."""
    model.eval()
    probs, labs, n_bad = [], [], 0
    for x, y, v in loader:
        x = x.to(DEVICE)
        if tta:
            views = [torch.rot90(torch.flip(x, dims=(3,)) if f else x, k, dims=(2, 3))
                     for f in (False, True) for k in range(4)]
        else:
            views = [x]
        acc = 0.0
        for xv in views:
            with torch.amp.autocast("cuda", enabled=(DEVICE.type == "cuda")):
                lg = model(xv)
            acc = acc + torch.sigmoid(lg.float())
        p, bad = _sanitize((acc / len(views)).cpu().numpy())
        n_bad += bad
        probs.append(p)
        labs.append(y.numpy())
    if n_bad:
        print(f"    [!] {n_bad} non-finite prediction(s) replaced")
    return np.concatenate(probs), np.concatenate(labs)

# ============================================================
# 5. MODEL (ResNet encoder, optionally SSL-initialised)
# ============================================================
ARCH_FEATS = {"resnet18": 512, "resnet34": 512, "resnet50": 2048}


def build_encoder(arch, in_ch=11, pretrained=True):
    ctor = getattr(torchvision.models, arch)
    wcls = getattr(torchvision.models, arch.replace("resnet", "ResNet") + "_Weights")
    net = ctor(weights=wcls.DEFAULT if pretrained else None)
    old = net.conv1
    new = nn.Conv2d(in_ch, 64, kernel_size=7, stride=2, padding=3, bias=False)
    with torch.no_grad():
        if pretrained:
            w_rgb = old.weight
            if CONFIG["bgr_to_rgb_init"]:
                w_rgb = w_rgb[:, [2, 1, 0]]          # data order is B,G,R -> map onto R,G,B filters
            new.weight[:, 0:3] = w_rgb
            mean_w = old.weight.mean(dim=1)
            for c in range(3, in_ch):
                new.weight[:, c] = mean_w * 0.25
        else:
            nn.init.kaiming_normal_(new.weight, nonlinearity="relu")
    net.conv1 = new
    net.fc = nn.Identity()
    return net


class ResNetClassifier(nn.Module):
    def __init__(self, arch, ssl_ckpt=None):
        super().__init__()
        use_ssl = ssl_ckpt is not None
        self.encoder = build_encoder(arch, in_ch=C, pretrained=not use_ssl)
        if use_ssl:
            if not ssl_ckpt.exists():
                raise FileNotFoundError(f"SSL checkpoint not found: {ssl_ckpt}")
            self.encoder.load_state_dict(torch.load(ssl_ckpt, map_location="cpu"))
            print("  [+] Loaded SSL weights")
        self.head = nn.Sequential(nn.Dropout(CONFIG["dropout"]), nn.Linear(ARCH_FEATS[arch], 1))

    def forward(self, x, valid=None):
        return self.head(self.encoder(x)).squeeze(1)


def build_optimizer(model):
    lr = CONFIG["lr"]
    return torch.optim.AdamW(
        [{"params": model.encoder.parameters(), "lr": lr * CONFIG["encoder_lr_mult"]},
         {"params": model.head.parameters(), "lr": lr}],
        weight_decay=CONFIG["weight_decay"],
    )

# ============================================================
# 6. REPORTING & PLOTS
# ============================================================
def print_fold_report(tag, fold, labs, probs):
    m = compute_metrics(labs, probs)
    print(f"\n  --- [{tag.upper()}] Fold {fold} validation report (final model, threshold 0.50) ---")
    print(f"    ROC-AUC: {m['auc']:.4f} | PR-AUC: {m['ap']:.4f} | F1: {m['f1']:.4f} | "
          f"Precision: {m['prec']:.4f} | Recall: {m['rec']:.4f} | Acc: {m['acc']:.4f}")
    print(f"    Confusion matrix -> TP: {m['tp']:<4} FP: {m['fp']:<4} FN: {m['fn']:<4} TN: {m['tn']:<4}")


def plot_fold_curves(tag, fold, h, val_labels, val_probs, best_ep):
    fig, axes = plt.subplots(2, 3, figsize=(18, 10))
    ep = h["epoch"]

    def decorate(ax, title, ylabel):
        ax.axvline(best_ep, color="gray", ls=":", lw=1.5, label=f"best epoch ({best_ep})")
        ax.set_title(f"{tag.upper()} Fold {fold}: {title}", fontweight="bold")
        ax.set_xlabel("Epoch"); ax.set_ylabel(ylabel)
        ax.grid(True, linestyle="--", alpha=0.6); ax.legend()

    ax = axes[0, 0]
    ax.plot(ep, h["trn_loss"], label="Train loss", color="tab:blue", lw=2)
    ax.plot(ep, h["val_loss"], label="Val loss", color="tab:orange", lw=2)
    decorate(ax, "Loss (true labels)", "BCE")

    ax = axes[0, 1]
    ax.plot(ep, h["trn_auc"], label="Train ROC-AUC", color="tab:blue", lw=2, ls="--")
    ax.plot(ep, h["val_auc"], label="Val ROC-AUC", color="tab:green", lw=2)
    ax.plot(ep, h["val_ap"], label="Val PR-AUC", color="tab:red", lw=2)
    decorate(ax, "AUC (train vs val)", "Score")

    ax = axes[0, 2]
    ax.plot(ep, h["val_prec"], label="Val precision", color="tab:purple", lw=2)
    ax.plot(ep, h["val_rec"], label="Val recall", color="tab:brown", lw=2)
    ax.plot(ep, h["val_f1"], label="Val F1", color="tab:red", lw=2)
    decorate(ax, "Val P / R / F1 (thr 0.5)", "Score")

    two = len(np.unique(val_labels)) == 2
    ax = axes[1, 0]
    if two:
        fpr, tpr, _ = roc_curve(val_labels, val_probs)
        ax.plot(fpr, tpr, label=f"ROC (AUC={roc_auc_score(val_labels, val_probs):.3f})", color="tab:purple", lw=2)
    ax.plot([0, 1], [0, 1], "k--", lw=1, alpha=0.5)
    ax.set_title(f"{tag.upper()} Fold {fold}: ROC (final model)", fontweight="bold")
    ax.set_xlabel("FPR"); ax.set_ylabel("TPR"); ax.grid(True, linestyle="--", alpha=0.6); ax.legend()

    ax = axes[1, 1]
    if two:
        prec, rec, _ = precision_recall_curve(val_labels, val_probs)
        ax.plot(rec, prec, label=f"PR (AP={average_precision_score(val_labels, val_probs):.3f})", color="tab:brown", lw=2)
        ax.axhline(val_labels.mean(), color="k", ls="--", lw=1, alpha=0.5, label="Prevalence")
    ax.set_title(f"{tag.upper()} Fold {fold}: PR (final model)", fontweight="bold")
    ax.set_xlabel("Recall"); ax.set_ylabel("Precision"); ax.grid(True, linestyle="--", alpha=0.6); ax.legend()

    ax = axes[1, 2]
    ax.plot(ep, h["lr"], color="tab:cyan", lw=2, label="Encoder LR")
    ax.set_yscale("log")
    decorate(ax, "Learning rate", "LR")

    plt.tight_layout()
    out_path = CONFIG["ckpt_dir"] / f"{tag}_fold{fold}_curves.png"
    plt.savefig(out_path, dpi=200)
    plt.show()
    plt.close(fig)
    print(f"    [+] Saved fold {fold} plots to: {out_path}")

# ============================================================
# 7. GROUP CROSS-VALIDATION PIPELINE
# ============================================================
def run_resnet_cv(ssl_ckpt, tag):
    arch, mon = CONFIG["arch"], CONFIG["monitor"]
    y_all = np.array([r["label"] for r in trainval_recs])
    g_all = pd.factorize([str(r["group"]) for r in trainval_recs])[0]

    sgkf = StratifiedGroupKFold(n_splits=CONFIG["n_folds"], shuffle=True, random_state=CONFIG["seed"])
    oof = np.zeros(len(trainval_recs))
    test_folds, fold_rows = [], []

    y_test = np.array([r["label"] for r in test_recs])
    test_ld = make_loader(test_recs, train=False)

    print(f"\n{'=' * 60}\n  Training {arch} ({tag.upper()}) | Stratified Group {CONFIG['n_folds']}-Fold "
          f"| monitor = val {mon.upper()}\n{'=' * 60}")

    for fold, (tr, va) in enumerate(sgkf.split(np.arange(len(trainval_recs)), y_all, g_all), 1):
        set_seed(CONFIG["seed"] + fold)          # same fold -> same RNG stream across tags
        tr_recs_f = [trainval_recs[i] for i in tr]
        va_recs_f = [trainval_recs[i] for i in va]

        tr_ld = make_loader(tr_recs_f, train=True)
        va_ld = make_loader(va_recs_f, train=False)

        model = ResNetClassifier(arch, ssl_ckpt=ssl_ckpt).to(DEVICE)
        crit = nn.BCEWithLogitsLoss()
        opt = build_optimizer(model)
        sched = torch.optim.lr_scheduler.ReduceLROnPlateau(opt, "max", factor=0.5, patience=CONFIG["sched_patience"])
        scaler = torch.amp.GradScaler("cuda", enabled=(DEVICE.type == "cuda"))

        h = {k: [] for k in ["epoch", "trn_loss", "val_loss", "trn_auc", "val_auc", "val_ap",
                             "val_f1", "val_prec", "val_rec", "lr"]}

        print(f"\n  --- Fold {fold}/{CONFIG['n_folds']} ({len(tr_recs_f)} train / {len(va_recs_f)} val patches) ---")

        best, best_ep, best_state, no_imp = -1.0, 0, None, 0
        for ep in range(1, CONFIG["epochs"] + 1):
            tr_m = run_epoch(model, tr_ld, crit, opt, scaler, train=True)
            va_m = run_epoch(model, va_ld, crit, train=False)

            score = va_m[mon]
            score = -1.0 if np.isnan(score) else score
            sched.step(score)

            h["epoch"].append(ep)
            h["trn_loss"].append(tr_m["loss"]); h["val_loss"].append(va_m["loss"])
            h["trn_auc"].append(tr_m["auc"]);   h["val_auc"].append(va_m["auc"])
            h["val_ap"].append(va_m["ap"]);     h["val_f1"].append(va_m["f1"])
            h["val_prec"].append(va_m["prec"]); h["val_rec"].append(va_m["rec"])
            h["lr"].append(opt.param_groups[0]["lr"])

            improved = score > best + 1e-4
            print(f"    Ep {ep:02d}/{CONFIG['epochs']} | "
                  f"Trn [L {tr_m['loss']:.4f} AUC {tr_m['auc']:.4f} P {tr_m['prec']:.3f} R {tr_m['rec']:.3f}] | "
                  f"Val [L {va_m['loss']:.4f} AUC {va_m['auc']:.4f} AP {va_m['ap']:.4f} "
                  f"P {va_m['prec']:.3f} R {va_m['rec']:.3f} F1 {va_m['f1']:.3f}] | "
                  f"lr {opt.param_groups[0]['lr']:.1e}"
                  f"{'  <- best' if improved else ''}")

            if improved:
                best, best_ep, no_imp = score, ep, 0
                best_state = {k: v.detach().cpu().clone() for k, v in model.state_dict().items()}
            else:
                no_imp += 1
                if no_imp >= CONFIG["early_stop_patience"]:
                    print(f"    Early stopping fold {fold} at epoch {ep} "
                          f"(no improvement for {CONFIG['early_stop_patience']} epochs)")
                    break

        if best_state is None:                    # metric never valid/improved: fall back to last weights
            print("    [!] No valid best epoch recorded — using final-epoch weights")
            best_state = {k: v.detach().cpu().clone() for k, v in model.state_dict().items()}
            best_ep = h["epoch"][-1]
        model.load_state_dict(best_state)

        # Final predictions (optionally with TTA), same procedure for OOF and test
        vp, vlab = predict_probs(model, va_ld, tta=CONFIG["use_tta"])
        oof[va] = vp
        tp_, _ = predict_probs(model, test_ld, tta=CONFIG["use_tta"])
        test_folds.append(tp_)

        print_fold_report(tag, fold, vlab, vp)
        plot_fold_curves(tag, fold, h, vlab, vp, best_ep)

        m = compute_metrics(vlab, vp)
        fold_rows.append({"fold": fold, "best_ep": best_ep, "auc": m["auc"], "ap": m["ap"], "f1": m["f1"]})
        print(f"  [{tag}] Fold {fold} done: best val {mon.upper()} = {best:.4f} at epoch {best_ep}")

        del model, opt, sched, scaler, tr_ld, va_ld
        gc.collect()
        if DEVICE.type == "cuda":
            torch.cuda.empty_cache()

    oof_auc = roc_auc_score(y_all, oof)
    ens = np.mean(test_folds, 0)

    print(f"\n[{tag}] Per-fold summary (final model on its val fold)")
    print("    fold | best_ep |  AUC   |   AP   |   F1")
    for r in fold_rows:
        print(f"    {r['fold']:>4} | {r['best_ep']:>7} | {r['auc']:.4f} | {r['ap']:.4f} | {r['f1']:.4f}")
    aucs = np.array([r["auc"] for r in fold_rows]); aps = np.array([r["ap"] for r in fold_rows])
    print(f"    mean AUC {np.nanmean(aucs):.4f} ± {np.nanstd(aucs):.4f} | mean AP {np.nanmean(aps):.4f} ± {np.nanstd(aps):.4f}")
    print(f"[{tag}] OOF ROC-AUC = {oof_auc:.4f} | TEST ens AUC = {roc_auc_score(y_test, ens):.4f}")

    np.savez(CONFIG["ckpt_dir"] / f"{tag}_preds.npz",
             oof=oof, oof_lab=y_all, test_ens=ens, test_lab=y_test, test_folds=np.array(test_folds))

    return {"tag": tag, "oof": oof, "oof_lab": y_all, "test_ens": ens,
            "test_lab": y_test, "oof_auc": oof_auc}

# ============================================================
# 8. EXECUTION, EVALUATION, AND PLOTTING
# ============================================================
def get_best_threshold_and_f1_curve(oof, lab, grid=THR_GRID):
    best_t, best_f1, f1_curve = 0.5, -1.0, []
    for t in grid:
        f = f1_score(lab, oof >= t, zero_division=0)
        f1_curve.append(f)
        if f > best_f1:
            best_f1, best_t = f, t
    return best_t, best_f1, f1_curve, grid


if __name__ == "__main__":
    results = []

    # NOTE: this run is ImageNet-initialised (not random init) -> tagged "imagenet"
    results.append(run_resnet_cv(ssl_ckpt=None, tag="imagenet"))

    if CONFIG["arch"] != CONFIG["ssl_arch"]:
        print(f"[!] Skipping SSL run: SSL checkpoint was trained with {CONFIG['ssl_arch']} "
              f"but CONFIG['arch'] = {CONFIG['arch']}")
    elif CONFIG["ssl_ckpt"].exists():
        results.append(run_resnet_cv(ssl_ckpt=CONFIG["ssl_ckpt"], tag="ssl"))
    else:
        print(f"[!] SSL checkpoint not found at {CONFIG['ssl_ckpt']} — skipping SSL comparison")

    print("\n" + "=" * 70)
    print("FINAL TEST-SET BENCHMARK EVALUATION")
    print("=" * 70)

    for res in results:
        lab, ens = res["test_lab"], res["test_ens"]
        oof, oof_lab = res["oof"], res["oof_lab"]

        tst_auc = roc_auc_score(lab, ens)
        tst_ap = average_precision_score(lab, ens)
        oof_ap = average_precision_score(oof_lab, oof)

        t, best_f1, _, _ = get_best_threshold_and_f1_curve(oof, oof_lab)
        if t >= THR_GRID[-1] - 1e-9 or t <= THR_GRID[0] + 1e-9:
            print(f"  [!] OOF-optimal threshold sits on the edge of the search grid ({t:.2f})")
        pred = (ens >= t).astype(int)

        tst_acc = accuracy_score(lab, pred)
        tst_f1 = f1_score(lab, pred, zero_division=0)

        print(f"\nMODEL: {res['tag'].upper()}")
        print(f"  • OOF Threshold   : {t:.2f} (OOF Peak F1: {best_f1:.4f})")
        print(f"  • Test ROC-AUC    : {tst_auc:.4f}  | OOF ROC-AUC: {res['oof_auc']:.4f}")
        print(f"  • Test PR-AUC     : {tst_ap:.4f}  | OOF PR-AUC : {oof_ap:.4f}")
        print(f"  • Test Accuracy   : {tst_acc:.4f}")
        print(f"  • Test F1-Score   : {tst_f1:.4f}")

        print("\n  --- Per-Class Breakdown ---")
        for cls, name in [(1, "site"), (0, "no_site")]:
            p = precision_score(lab, pred, pos_label=cls, zero_division=0)
            r = recall_score(lab, pred, pos_label=cls, zero_division=0)
            print(f"    {name:8s} -> Precision: {p:.3f} | Recall: {r:.3f}")

        tn, fp, fn, tp = confusion_matrix(lab, pred, labels=[0, 1]).ravel()
        print("\n  --- Confusion Matrix ---")
        print(f"    TP: {tp:<4} FP: {fp:<4}")
        print(f"    FN: {fn:<4} TN: {tn:<4}")

        print("\n  --- Survey Prioritization (Recall @ Top-k%) ---")
        order = np.argsort(-ens)
        n_pos_test = int(lab.sum())
        for k in [0.10, 0.20, 0.30]:
            n = max(1, int(k * len(ens)))
            hit = int(lab[order[:n]].sum())
            rec_k = hit / n_pos_test if n_pos_test else float("nan")
            print(f"    Top {int(k * 100):2d}% ({n:3d} patches): Caught {hit}/{n_pos_test} sites (Recall = {rec_k:.3f})")
        print("-" * 70)

    fig, axes = plt.subplots(1, 3, figsize=(18, 5))
    colors = {"imagenet": "tab:blue", "ssl": "tab:orange"}

    for res in results:
        tag = res["tag"]
        c = colors.get(tag, "tab:green")
        lab, ens = res["test_lab"], res["test_ens"]
        oof, oof_lab = res["oof"], res["oof_lab"]

        fpr, tpr, _ = roc_curve(lab, ens)
        axes[0].plot(fpr, tpr, label=f"{tag.upper()} Test (AUC = {roc_auc_score(lab, ens):.3f})", color=c, lw=2)

        prec_c, rec_c, _ = precision_recall_curve(lab, ens)
        axes[1].plot(rec_c, prec_c, label=f"{tag.upper()} Test (AP = {average_precision_score(lab, ens):.3f})", color=c, lw=2)

        t, _, _, grid = get_best_threshold_and_f1_curve(oof, oof_lab)
        tst_f1_curve = [f1_score(lab, (ens >= g).astype(int), zero_division=0) for g in grid]
        axes[2].plot(grid, tst_f1_curve, label=f"{tag.upper()} Test F1", color=c, lw=2)
        axes[2].axvline(x=t, color=c, linestyle="--", lw=1.5, alpha=0.7, label=f"{tag.upper()} OOF-opt thr ({t:.2f})")

    axes[0].plot([0, 1], [0, 1], "k--", lw=1.5, label="Chance")
    axes[0].set_title("Test ROC Curves", fontweight="bold")
    axes[0].set_xlabel("False Positive Rate"); axes[0].set_ylabel("True Positive Rate")

    axes[1].set_title("Test Precision-Recall Curves", fontweight="bold")
    axes[1].set_xlabel("Recall"); axes[1].set_ylabel("Precision")

    axes[2].set_title("Test F1 Score vs Threshold", fontweight="bold")
    axes[2].set_xlabel("Classification Threshold"); axes[2].set_ylabel("F1 Score")
    axes[2].set_xlim([THR_GRID[0], THR_GRID[-1]]); axes[2].set_ylim([0.0, 1.02])

    for ax in axes:
        ax.grid(True, linestyle="--", alpha=0.6)
        ax.legend()

    plt.tight_layout()
    plot_path = CONFIG["ckpt_dir"] / "resnet_final_evaluation.png"
    plt.savefig(plot_path, dpi=300)
    print(f"\n[+] Saved diagnostic evaluation plots to: {plot_path.resolve()}")
    plt.show()

    gc.collect()
    if DEVICE.type == "cuda":
        torch.cuda.empty_cache()
```
