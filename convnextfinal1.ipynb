```python
# !pip install -q rasterio

import os, random, warnings, gc
from pathlib import Path
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader, WeightedRandomSampler
import torchvision.transforms.functional as TF
import torchvision.models as models
from sklearn.metrics import (
    roc_auc_score, average_precision_score,
    precision_score, recall_score, f1_score, accuracy_score,
    precision_recall_curve
)
import rasterio

warnings.filterwarnings('ignore')
DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print('Using device:', DEVICE)
torch.backends.cudnn.benchmark = True

# ══════════════════════════════════════════════════════════════════════════════
# CONFIGURATION
# ══════════════════════════════════════════════════════════════════════════════
class CFG:
    ROOT          = Path('/kaggle/input/datasets/shreyansdeshpande/200x200/FINALCNNDATA_200')
    OUTPUT_DIR    = Path('/kaggle/working')
    
    TRAIN_SITE    = ROOT / 'train' / 'site'
    TRAIN_NOSITE  = ROOT / 'train' / 'no_site'
    VAL_SITE      = ROOT / 'validation' / 'site'
    VAL_NOSITE    = ROOT / 'validation' / 'no_site'
    TEST_SITE     = ROOT / 'test' / 'site'
    TEST_NOSITE   = ROOT / 'test' / 'no_site'

    IMG_SUFFIX    = '.tif'
    SAT_BANDS     = [0, 1, 2, 3]
    LIDAR_BANDS   = [4, 5, 6, 7, 8, 9, 10]
    IN_CHANNELS   = len(SAT_BANDS) + len(LIDAR_BANDS)

    NUM_CLASSES   = 1
    IMG_SIZE      = 200
    BATCH_SIZE    = 32
    NUM_EPOCHS    = 30
    T_MAX         = NUM_EPOCHS
    ETA_MIN       = 1e-6
    BASE_LR       = 1e-4
    WEIGHT_DECAY  = 0.05
    DROPOUT       = 0.6
    SEED          = 42
    NUM_WORKERS   = 2
    USE_AMP       = torch.cuda.is_available()

def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

set_seed(CFG.SEED)
CFG.OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

# ══════════════════════════════════════════════════════════════════════════════
# DATA INGESTION & NORMALIZATION
# ══════════════════════════════════════════════════════════════════════════════
def folder_to_df(site_dir: Path, nosite_dir: Path) -> pd.DataFrame:
    rows = [{'filepath': str(f), 'label': l} for l, d in [(1, site_dir), (0, nosite_dir)] for f in sorted(d.glob(f'*{CFG.IMG_SUFFIX}'))]
    return pd.DataFrame(rows)

print('Scanning dataset folders...')
train_df = folder_to_df(CFG.TRAIN_SITE, CFG.TRAIN_NOSITE)
val_df   = folder_to_df(CFG.VAL_SITE,   CFG.VAL_NOSITE)
test_df  = folder_to_df(CFG.TEST_SITE,  CFG.TEST_NOSITE)

print(f"  Train : {len(train_df):<5} (site: {(train_df['label']==1).sum()}, no_site: {(train_df['label']==0).sum()})")
print(f"  Val   : {len(val_df):<5} (site: {(val_df['label']==1).sum()}, no_site: {(val_df['label']==0).sum()})")
print(f"  Test  : {len(test_df):<5} (site: {(test_df['label']==1).sum()}, no_site: {(test_df['label']==0).sum()})")

def normalize_lidar(arr: np.ndarray) -> np.ndarray:
    out = np.empty_like(arr, dtype=np.float32)
    for c, ch in enumerate(arr):
        mean, std = float(np.nanmean(ch)), float(np.nanstd(ch))
        out[c] = ch - mean if std < 1e-4 or np.isnan(std) else (ch - mean) / (std + 1e-6)
    return np.nan_to_num(out, nan=0.0, posinf=0.0, neginf=0.0)

def normalize_satellite(arr: np.ndarray) -> np.ndarray:
    arr = np.nan_to_num(arr, nan=0.0, posinf=0.0, neginf=0.0).clip(0.0)
    max_val = float(np.max(arr))
    if max_val > 255.0:
        return np.power(np.clip((arr / 10000.0) * 2.5, 0.0, 1.0), 1.0 / 2.2).astype(np.float32)
    return (arr / 255.0 if max_val > 1.0 else arr).astype(np.float32)

def load_multimodal_patch(tif_path: str) -> np.ndarray:
    with rasterio.open(tif_path) as src:
        img = src.read().astype(np.float32)
    img = np.nan_to_num(img, nan=0.0, posinf=0.0, neginf=0.0)
    img = np.where((img < -1e4) | (img > 1e4), 0.0, img)
    return np.concatenate([normalize_satellite(img[CFG.SAT_BANDS]), normalize_lidar(img[CFG.LIDAR_BANDS])], axis=0)

class MultiModalDataset(Dataset):
    def __init__(self, df: pd.DataFrame, split_name: str, is_train: bool = True):
        self.df = df.reset_index(drop=True)
        self.is_train = is_train
        print(f"Caching {len(self.df)} {split_name} patches into RAM...")
        self.cached = [(load_multimodal_patch(row.filepath), float(row.label)) for _, row in self.df.iterrows()]

    def __len__(self): return len(self.cached)

    def __getitem__(self, idx):
        img, label = self.cached[idx]
        t = TF.resize(torch.from_numpy(img), [CFG.IMG_SIZE, CFG.IMG_SIZE], interpolation=TF.InterpolationMode.BILINEAR, antialias=True)
        
        if self.is_train:
            if random.random() > 0.5: t = TF.hflip(t)
            if random.random() > 0.5: t = TF.vflip(t)
            if (angle := random.choice([0, 90, 180, 270])) != 0: t = TF.rotate(t, angle)
            if random.random() > 0.5: t = t + torch.randn_like(t) * 0.03
            
        return t, torch.tensor(label, dtype=torch.float32)

# ══════════════════════════════════════════════════════════════════════════════
# DATALOADERS & MODEL
# ══════════════════════════════════════════════════════════════════════════════
counts = train_df['label'].value_counts().to_dict()
sample_weights = [1.0 / counts[lbl] for lbl in train_df['label']]
sampler = WeightedRandomSampler(weights=sample_weights, num_samples=len(train_df), replacement=True)

trn_loader  = DataLoader(MultiModalDataset(train_df, "Train", True), batch_size=CFG.BATCH_SIZE, sampler=sampler, num_workers=CFG.NUM_WORKERS, pin_memory=True, drop_last=True)
val_loader  = DataLoader(MultiModalDataset(val_df, "Val", False), batch_size=CFG.BATCH_SIZE * 2, shuffle=False, num_workers=CFG.NUM_WORKERS, pin_memory=True)
test_loader = DataLoader(MultiModalDataset(test_df, "Test", False), batch_size=CFG.BATCH_SIZE * 2, shuffle=False, num_workers=CFG.NUM_WORKERS, pin_memory=True)

def create_convnext_model(in_channels=11, num_classes=1, dropout=0.6):
    # ── Step 1: Load pretrained ConvNeXt-Tiny (ImageNet weights) ─────────────
    model = models.convnext_tiny(weights=models.ConvNeXt_Tiny_Weights.DEFAULT)

    # ── `Step 2: Grab the original stem layer (4x4 conv, 3 -> 96 channels) ────
    old_conv = model.features[0][0]

    old_in_channels  = old_conv.weight.shape[1]      # normally 3 (RGB)
    old_out_channels = old_conv.out_channels          # normally 96
    kernel_size      = old_conv.kernel_size
    stride           = old_conv.stride
    padding          = old_conv.padding
    has_bias         = old_conv.bias is not None

    # ── Step 3: Build a brand new stem conv with our custom channel count ────
    new_conv = nn.Conv2d(
        in_channels=in_channels,
        out_channels=old_out_channels,
        kernel_size=kernel_size,
        stride=stride,
        padding=padding,
        bias=has_bias
    )

    # ── Step 4: Fill in the new conv's weights ────────────────────────────────
    with torch.no_grad():

        needs_channel_expansion = (old_in_channels == 3) and (in_channels > 3)

        if in_channels == old_in_channels:
    # No adaptation needed at all — just copy the pretrained weights directly
                new_conv.weight[:] = old_conv.weight

        elif old_in_channels == 3 and in_channels > 3:
    # Your existing expansion logic
              new_conv.weight[:, 0:3, :, :] = old_conv.weight
              mean_weight = old_conv.weight.mean(dim=1)
              for c in range(3, in_channels):
                   new_conv.weight[:, c, :, :] = mean_weight

        else:
    # Genuinely no sensible pretrained mapping (e.g. fewer channels than pretrained, or unusual shape)
             nn.init.kaiming_normal_(new_conv.weight, nonlinearity='relu')

        # 4d: Carry over the bias term unchanged, if one exists
        if has_bias:
            new_conv.bias.copy_(old_conv.bias)

    # ── Step 5: Swap the new stem conv into the model ─────────────────────────
    model.features[0][0] = new_conv

    # ── Step 6: Replace the classifier head for binary classification ────────
    in_features = model.classifier[2].in_features   # 768, ConvNeXt-Tiny's final channel width

    original_norm_layer  = model.classifier[0]        # LayerNorm2d (kept as-is)
    original_flatten_layer = model.classifier[1]       # Flatten (kept as-is)

    model.classifier = nn.Sequential(
        original_norm_layer,
        original_flatten_layer,
        nn.Dropout(p=dropout),
        nn.Linear(in_features, num_classes)
    )

    return model

# ══════════════════════════════════════════════════════════════════════════════
# LOSS, OPTIMIZER & METRICS HELPER
# ══════════════════════════════════════════════════════════════════════════════
model = create_convnext_model(CFG.IN_CHANNELS, CFG.NUM_CLASSES, CFG.DROPOUT).to(DEVICE)
optimizer = optim.AdamW(model.parameters(), lr=CFG.BASE_LR, weight_decay=CFG.WEIGHT_DECAY)
scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=CFG.T_MAX, eta_min=CFG.ETA_MIN)
scaler    = torch.amp.GradScaler(device=DEVICE.type, enabled=CFG.USE_AMP)
criterion = nn.BCEWithLogitsLoss()

best_val_auc = 0.0
ckpt_path    = CFG.OUTPUT_DIR / 'convnext_tiny_multimodal_best.pth'

def compute_metrics(y_true, y_pred):
    auc = roc_auc_score(y_true, y_pred) if len(set(y_true)) > 1 else 0.0
    pr_auc = average_precision_score(y_true, y_pred) if len(set(y_true)) > 1 else 0.0
    bin_preds = [1 if p >= 0.5 else 0 for p in y_pred]
    return auc, pr_auc, precision_score(y_true, bin_preds, zero_division=0), recall_score(y_true, bin_preds, zero_division=0), f1_score(y_true, bin_preds, zero_division=0), accuracy_score(y_true, bin_preds)

def train_epoch(model, loader, optimizer, scaler):
    model.train()
    total_loss, all_labels, all_preds = 0.0, [], []
    for inputs, labels in loader:
        inputs, labels = inputs.to(DEVICE, non_blocking=True), labels.to(DEVICE, non_blocking=True)
        optimizer.zero_grad(set_to_none=True)
        with torch.amp.autocast(device_type=DEVICE.type, enabled=CFG.USE_AMP):
            loss = criterion(logits := model(inputs).squeeze(1), labels)
        scaler.scale(loss).backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        scaler.step(optimizer); scaler.update()
        total_loss += loss.item() * labels.size(0)
        all_preds.extend(np.nan_to_num(torch.sigmoid(logits).detach().cpu().numpy(), nan=0.5))
        all_labels.extend(labels.cpu().numpy())
    return total_loss / len(loader.dataset), *compute_metrics(all_labels, all_preds)

@torch.no_grad()
def evaluate(model, loader):
    model.eval()
    total_loss, all_labels, all_preds = 0.0, [], []
    for inputs, labels in loader:
        inputs, labels = inputs.to(DEVICE, non_blocking=True), labels.to(DEVICE, non_blocking=True)
        with torch.amp.autocast(device_type=DEVICE.type, enabled=CFG.USE_AMP):
            loss = criterion(logits := model(inputs).squeeze(1), labels)
        total_loss += loss.item() * labels.size(0)
        all_preds.extend(np.nan_to_num(torch.sigmoid(logits).cpu().numpy(), nan=0.5))
        all_labels.extend(labels.cpu().numpy())
    return total_loss / len(loader.dataset), *compute_metrics(all_labels, all_preds), all_labels, all_preds

# ══════════════════════════════════════════════════════════════════════════════
# TRAINING LOOP & EVALUATION
# ══════════════════════════════════════════════════════════════════════════════
print(f'\n{"="*75}\n  Training ConvNeXt-Tiny on 11-Channel Multimodal Data ({CFG.NUM_EPOCHS} Epochs)\n{"="*75}')

for epoch in range(1, CFG.NUM_EPOCHS + 1):
    trn_loss, trn_auc, trn_pr_auc, trn_prec, trn_rec, trn_f1, trn_acc = train_epoch(model, trn_loader, optimizer, scaler)
    val_loss, val_auc, val_pr_auc, val_prec, val_rec, val_f1, val_acc, _, _ = evaluate(model, val_loader)
    scheduler.step()

    saved = ''
    if val_auc > best_val_auc:
        best_val_auc = val_auc
        torch.save(model.state_dict(), ckpt_path)
        saved = '  ✓ saved'

    print(f'Ep {epoch:02d}/{CFG.NUM_EPOCHS} | LR: {optimizer.param_groups[0]["lr"]:.2e}{saved}')
    print(f'  Train -> Loss: {trn_loss:.4f} | ROC-AUC: {trn_auc:.4f} | PR-AUC: {trn_pr_auc:.4f} | F1: {trn_f1:.4f}')
    print(f'  Val   -> Loss: {val_loss:.4f} | ROC-AUC: {val_auc:.4f} | PR-AUC: {val_pr_auc:.4f} | F1: {val_f1:.4f}')

if ckpt_path.exists():
    model.load_state_dict(torch.load(ckpt_path, map_location=DEVICE))
    model.eval()

    _, _, _, _, _, _, _, val_labels, val_preds = evaluate(model, val_loader)
    precisions, recalls, thresholds = precision_recall_curve(val_labels, val_preds)
    best_thresh = float(thresholds[np.argmax(2 * (precisions[:-1] * recalls[:-1]) / (precisions[:-1] + recalls[:-1] + 1e-8))])

    _, tst_auc, tst_pr_auc, _, _, _, _, test_labels, test_preds = evaluate(model, test_loader)
    test_preds_np, test_labels_np = np.array(test_preds), np.array(test_labels)

    bin_default, bin_opt = (test_preds_np >= 0.5).astype(int), (test_preds_np >= best_thresh).astype(int)

    print("\n" + "=" * 65 + "\n      FINAL BENCHMARK: UNTOUCHED TEST SET (CONVNEXT-TINY)\n" + "=" * 65)
    print(f"Optimal Threshold Calibrated on Val : {best_thresh:.4f}")
    print(f"\nTest Metrics @ Default Threshold (0.50):\n  • ROC-AUC     : {tst_auc:.4f}\n  • PR-AUC      : {tst_pr_auc:.4f}\n  • F1-Score    : {f1_score(test_labels_np, bin_default, zero_division=0):.4f}")
    print(f"\nTest Metrics @ Optimal Calibrated Threshold ({best_thresh:.4f}):\n  • ROC-AUC     : {tst_auc:.4f}\n  • PR-AUC      : {tst_pr_auc:.4f}\n  • F1-Score    : {f1_score(test_labels_np, bin_opt, zero_division=0):.4f}")
    print("=" * 65)

gc.collect()
torch.cuda.empty_cache()
```
