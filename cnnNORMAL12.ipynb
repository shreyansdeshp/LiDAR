```python
# -*- coding: utf-8 -*-
"""
FINAL LEARNING-CURVE SCRIPT — WITH 5-FOLD SPATIAL CV PER FRACTION
============================================================
For each fraction (30% / 40% / 50% / 60% / 70%) of the trainval
pool:

  1. A NESTED, STRATIFIED subset is taken (30% subset ⊂ 40%
     subset ⊂ ... ⊂ 70% subset; site/no_site ratio preserved).
  2. That subset is split into a BALANCED SPATIAL 5-FOLD CV
     (StratifiedGroupKFold, searched over many seeds for the
     most balanced split, with a hard check that no spatial
     group ever appears in both train and val for a fold).
  3. 5 models are trained, one per fold, each early-stopping on
     ITS OWN fold validation set (the frozen test set is never
     touched during training or model selection).
  4. A decision threshold is locked from the 5 folds' combined
     out-of-fold (OOF) predictions (test set still untouched).
  5. The 5 fold models are ENSEMBLED (mean probability) and
     evaluated ONCE on the FROZEN test set.

The test set (test_fp / test_labels) is loaded once, at the top,
and is never resampled, retrained on, or used for any decision
(threshold, early stopping, model selection) — only for the
final, single evaluation per fraction.

Validation data is merged into train (no separate held-out
validation set exists anywhere in this script) — the 5 folds
above are carved dynamically out of that merged trainval pool.

All results are saved to disk under CONFIG["output_dir"].
============================================================
"""

# ============================================================
# SECTION 0 — IMPORTS + REPRODUCIBILITY + DEVICE
# ============================================================

import os
import glob
import random
import copy

import numpy as np
import pandas as pd
import rasterio
import matplotlib.pyplot as plt

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader, WeightedRandomSampler

from sklearn.model_selection import StratifiedGroupKFold
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
)

from tqdm.auto import tqdm

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("=" * 80)
print("DEVICE")
print("=" * 80)
print("PyTorch:", torch.__version__)
print("Device :", DEVICE)
print("=" * 80)


# ============================================================
# SECTION 1 — CONFIG
# ============================================================

CONFIG = {
    "n_channels": 11,
    "n_handcrafted": 15,
    "patch_size": 250,

    "dropout": 0.35,
    "use_se_blocks": True,
    "use_handcrafted_features": True,

    "use_channel_dropout": True,
    "channel_dropout_prob": 0.5,
    "channel_dropout_max_channels": 2,

    "use_balanced_sampler": True,

    "batch_size": 16,
    "epochs": 30,
    "learning_rate": 3e-4,
    "weight_decay": 1e-4,
    "grad_clip_norm": 1.0,

    "scheduler_factor": 0.5,
    "scheduler_patience": 3,

    "early_stopping_patience": 8,

    # --------------------------------------------------------
    # 5-FOLD CV SETTINGS
    # --------------------------------------------------------
    "n_folds": 5,
    "fold_search_seeds": 200,   # how many seeds to try when
                                 # searching for the most balanced
                                 # spatial 5-fold split

    # --------------------------------------------------------
    # DATA LOCATION — confirmed real path
    # --------------------------------------------------------
    "data_root": (
        "/kaggle/input/datasets/shreyansdeshpande/"
        "2nd-variation/FINALCNNDATA_NEW_01"
    ),
    "metadata_path": (
        "/kaggle/input/datasets/shreyansdeshpande/"
        "2nd-variation/FINALCNNDATA_NEW_01/"
        "metadata/combined/metadata_all.csv"
    ),

    # --------------------------------------------------------
    # OUTPUT LOCATION — everything gets saved here
    # --------------------------------------------------------
    "output_dir": "/kaggle/working/learning_curve_5fold_results",

    # --------------------------------------------------------
    # LEARNING CURVE SETTINGS
    # --------------------------------------------------------
    "fractions": [0.30, 0.50, 0.70],
}

os.makedirs(CONFIG["output_dir"], exist_ok=True)

print("\n" + "=" * 80)
print("CONFIG")
print("=" * 80)
for k, v in CONFIG.items():
    print(f"{k:20s}: {v}")
print("=" * 80)


# ============================================================
# SECTION 2 — LOAD TRAIN / VALIDATION / TEST, MERGE TRAIN+VAL
# ============================================================

DATA_ROOT = CONFIG["data_root"]
METADATA_PATH = CONFIG["metadata_path"]


def collect_split(split_name):
    split_root = os.path.join(DATA_ROOT, split_name)
    site_dir = os.path.join(split_root, "site")
    no_site_dir = os.path.join(split_root, "no_site")

    site_files = sorted(
        glob.glob(os.path.join(site_dir, "*.tif")) +
        glob.glob(os.path.join(site_dir, "*.tiff"))
    )
    no_site_files = sorted(
        glob.glob(os.path.join(no_site_dir, "*.tif")) +
        glob.glob(os.path.join(no_site_dir, "*.tiff"))
    )

    filepaths = site_files + no_site_files
    labels = np.array(
        [1] * len(site_files) + [0] * len(no_site_files),
        dtype=np.int64,
    )
    return filepaths, labels


print("\n" + "=" * 80)
print("LOADING SPLITS")
print("=" * 80)

train_fp, train_labels = collect_split("train")
val_fp, val_labels = collect_split("validation")
test_fp, test_labels = collect_split("test")

# ------------------------------------------------------------
# MERGE TRAIN + VALIDATION  ->  trainval
# (validation is NOT used as a separate held-out set anywhere
#  downstream — it becomes part of the pool that gets 5-folded)
# ------------------------------------------------------------

trainval_fp = train_fp + val_fp
trainval_labels = np.concatenate([train_labels, val_labels])

print(f"Train samples       : {len(train_fp)}")
print(f"Validation samples  : {len(val_fp)}  (merged into train)")
print(f"Train+Val samples   : {len(trainval_fp)}")
print(f"TEST samples        : {len(test_fp)}  (FROZEN, never used until final eval)")
print("=" * 80)


# ============================================================
# SECTION 3 — LOAD METADATA: MONUMENT COUNTS + SPATIAL GROUPS
# ============================================================

metadata_df = pd.read_csv(METADATA_PATH)

# ------------------------------------------------------------
# Find patch-name / label / monument-count / spatial-group columns
# ------------------------------------------------------------

def find_column(candidates, columns):
    for c in candidates:
        if c in columns:
            return c
    raise KeyError(
        f"None of {candidates} found in columns: {list(columns)}"
    )


PATCH_COL = find_column(
    ["patch_name", "patch", "filename", "file_name", "tif_name", "name"],
    metadata_df.columns,
)
MONUMENT_COUNT_COL = find_column(
    ["monument_count", "monument_counts", "n_monuments",
     "num_monuments", "number_of_monuments"],
    metadata_df.columns,
)
GROUP_COL = find_column(
    ["spatial_group_id", "spatial_group", "group_id", "group"],
    metadata_df.columns,
)

print("\nMetadata columns identified:")
print(f"  Patch column          : {PATCH_COL}")
print(f"  Monument count column : {MONUMENT_COUNT_COL}")
print(f"  Spatial group column  : {GROUP_COL}")

metadata_df["_basename"] = (
    metadata_df[PATCH_COL].astype(str).apply(os.path.basename)
)
metadata_df[MONUMENT_COUNT_COL] = (
    pd.to_numeric(metadata_df[MONUMENT_COUNT_COL], errors="coerce")
    .fillna(0).astype(np.int64)
)

monument_count_lookup = dict(
    zip(metadata_df["_basename"], metadata_df[MONUMENT_COUNT_COL])
)
group_lookup = dict(
    zip(metadata_df["_basename"], metadata_df[GROUP_COL])
)

# ------------------------------------------------------------
# Align monument counts + spatial groups to trainval / test
# ------------------------------------------------------------

trainval_monument_counts = np.array(
    [monument_count_lookup.get(os.path.basename(fp), 0) for fp in trainval_fp],
    dtype=np.int64,
)
test_monument_counts = np.array(
    [monument_count_lookup.get(os.path.basename(fp), 0) for fp in test_fp],
    dtype=np.int64,
)

missing_groups = [
    os.path.basename(fp) for fp in trainval_fp
    if os.path.basename(fp) not in group_lookup
]
if missing_groups:
    raise RuntimeError(
        f"{len(missing_groups)} trainval patches have no spatial "
        f"group in metadata. First few: {missing_groups[:10]}"
    )

trainval_groups = np.array(
    [group_lookup[os.path.basename(fp)] for fp in trainval_fp]
)

print("\nMetadata rows:", len(metadata_df))
print("Trainval monument counts aligned:", len(trainval_monument_counts))
print("Trainval spatial groups aligned :", len(trainval_groups))
print("Unique spatial groups in trainval:", len(np.unique(trainval_groups)))


# ============================================================
# SECTION 4 — SOFT LABELS (monument count -> soft target)
# ============================================================

SOFT_LABEL_MAP = {
    0: 0.05, 1: 0.7377, 2: 0.8212, 3: 0.8719, 4: 0.9026,
    5: 0.9213, 6: 0.9315, 7: 0.9380, 8: 0.9425, 9: 0.9455, 10: 0.9476,
}


def monument_count_to_soft_label(count):
    count = int(count)
    if count <= 0:
        return 0.05
    if count >= 20:
        return 0.95
    if count in SOFT_LABEL_MAP:
        return SOFT_LABEL_MAP[count]
    return float(np.interp(count, [10, 20], [0.9476, 0.95]))


# ============================================================
# SECTION 5 — RASTER LOADER + HANDCRAFTED FEATURES
# ============================================================

def load_patch(filepath):
    if filepath.lower().endswith(".npy"):
        arr = np.load(filepath).astype(np.float32)
    else:
        with rasterio.open(filepath) as src:
            arr = src.read().astype(np.float32)

    arr = np.nan_to_num(arr, nan=0.0, posinf=0.0, neginf=0.0)

    expected_shape = (
        CONFIG["n_channels"], CONFIG["patch_size"], CONFIG["patch_size"]
    )
    if arr.shape != expected_shape:
        raise ValueError(
            f"\nUnexpected patch shape.\nFile: {filepath}\n"
            f"Got: {arr.shape}\nExpected: {expected_shape}"
        )
    return arr


def compute_handcrafted_features(arr):
    LRM = arr[0]
    SVF = arr[1]
    SLOPE = arr[2]
    LOCAL_DOM = arr[10]
    R = arr[6]
    NIR = arr[9]

    feats = []
    feats += [LRM.mean(), LRM.std(), LRM.min(), LRM.max(), LRM.max() - LRM.min()]

    gy, gx = np.gradient(LRM)
    edge_mag = np.sqrt(gx ** 2 + gy ** 2)
    feats += [edge_mag.mean(), edge_mag.std()]

    feats += [SLOPE.mean(), SLOPE.max()]
    feats += [SVF.mean(), SVF.std()]
    feats += [LOCAL_DOM.mean(), LOCAL_DOM.std()]

    ndvi = (NIR - R) / (NIR + R + 1e-6)
    feats += [ndvi.mean(), ndvi.std()]

    return np.asarray(feats, dtype=np.float32)


def compute_channel_stats(filepaths, max_samples=800):
    sampled_files = (
        filepaths if len(filepaths) <= max_samples
        else random.sample(filepaths, max_samples)
    )
    n_channels = CONFIG["n_channels"]
    channel_sum = np.zeros(n_channels, dtype=np.float64)
    channel_sum_sq = np.zeros(n_channels, dtype=np.float64)
    total_pixels = 0

    for filepath in tqdm(sampled_files, desc="Computing channel statistics"):
        arr = load_patch(filepath)
        pixels = arr.reshape(n_channels, -1)
        channel_sum += pixels.sum(axis=1)
        channel_sum_sq += (pixels ** 2).sum(axis=1)
        total_pixels += pixels.shape[1]

    mean = channel_sum / total_pixels
    variance = np.maximum(channel_sum_sq / total_pixels - mean ** 2, 1e-8)
    std = np.sqrt(variance)
    std[std < 1e-6] = 1.0
    return mean.astype(np.float32), std.astype(np.float32)


def compute_handcrafted_stats(filepaths, max_samples=800):
    sampled_files = (
        filepaths if len(filepaths) <= max_samples
        else random.sample(filepaths, max_samples)
    )
    all_features = np.zeros(
        (len(sampled_files), CONFIG["n_handcrafted"]), dtype=np.float64
    )
    for i, filepath in enumerate(tqdm(sampled_files, desc="Computing handcrafted statistics")):
        arr = load_patch(filepath)
        all_features[i] = compute_handcrafted_features(arr)

    mean = all_features.mean(axis=0)
    std = all_features.std(axis=0)
    std[std < 1e-6] = 1.0
    return mean.astype(np.float32), std.astype(np.float32)


print("\n" + "=" * 80)
print("COMPUTING NORMALIZATION STATISTICS (from full trainval pool)")
print("=" * 80)

CHANNEL_MEAN, CHANNEL_STD = compute_channel_stats(trainval_fp)
HANDCRAFTED_MEAN, HANDCRAFTED_STD = compute_handcrafted_stats(trainval_fp)


# ============================================================
# SECTION 6 — DATASET CLASS
# ============================================================

class TerrainPatchDataset(Dataset):

    def __init__(self, filepaths, hard_labels, monument_counts,
                 mean, std, hc_mean, hc_std,
                 augment=False, use_handcrafted=True,
                 use_channel_dropout=True, channel_dropout_prob=0.5,
                 channel_dropout_max_channels=2):

        self.filepaths = filepaths
        self.hard_labels = np.asarray(hard_labels, dtype=np.int64)
        self.monument_counts = np.asarray(monument_counts, dtype=np.int64)
        self.mean = mean.reshape(-1, 1, 1)
        self.std = std.reshape(-1, 1, 1)
        self.hc_mean = hc_mean
        self.hc_std = hc_std
        self.augment = augment
        self.use_handcrafted = use_handcrafted
        self.use_channel_dropout = use_channel_dropout
        self.channel_dropout_prob = channel_dropout_prob
        self.channel_dropout_max_channels = channel_dropout_max_channels

    def __len__(self):
        return len(self.filepaths)

    @staticmethod
    def _geometric_augment(arr):
        if random.random() < 0.5:
            arr = arr[:, :, ::-1]
        if random.random() < 0.5:
            arr = arr[:, ::-1, :]
        k = random.choice([0, 1, 2, 3])
        if k != 0:
            arr = np.rot90(arr, k=k, axes=(1, 2))
        return arr

    def _channel_dropout(self, arr):
        if random.random() < self.channel_dropout_prob:
            n_drop = random.randint(1, self.channel_dropout_max_channels)
            drop_idx = random.sample(range(arr.shape[0]), n_drop)
            arr[drop_idx] = 0.0
        return arr

    def __getitem__(self, idx):
        filepath = self.filepaths[idx]
        hard_label = int(self.hard_labels[idx])
        monument_count = int(self.monument_counts[idx])

        raw_arr = load_patch(filepath)

        if self.use_handcrafted:
            hc_feat = compute_handcrafted_features(raw_arr)
            hc_feat = (hc_feat - self.hc_mean) / self.hc_std
        else:
            hc_feat = np.zeros(1, dtype=np.float32)

        arr = raw_arr
        if self.augment:
            arr = self._geometric_augment(arr)
        arr = np.ascontiguousarray(arr)

        arr = (arr - self.mean) / self.std

        if self.augment and self.use_channel_dropout:
            arr = self._channel_dropout(arr)

        soft_label = monument_count_to_soft_label(monument_count)

        return (
            torch.from_numpy(arr).float(),
            torch.from_numpy(hc_feat).float(),
            torch.tensor(soft_label, dtype=torch.float32),
            torch.tensor(hard_label, dtype=torch.long),
        )


# ============================================================
# SECTION 7 — MODEL
# ============================================================

class ConvBNReLU(nn.Module):
    def __init__(self, in_ch, out_ch, kernel_size=3, padding=1):
        super().__init__()
        self.conv = nn.Conv2d(in_ch, out_ch, kernel_size, padding=padding, bias=False)
        self.bn = nn.BatchNorm2d(out_ch)
        self.relu = nn.ReLU(inplace=True)

    def forward(self, x):
        return self.relu(self.bn(self.conv(x)))


class SEBlock(nn.Module):
    def __init__(self, channels, reduction=8):
        super().__init__()
        reduced = max(channels // reduction, 4)
        self.pool = nn.AdaptiveAvgPool2d(1)
        self.fc1 = nn.Linear(channels, reduced)
        self.relu = nn.ReLU(inplace=True)
        self.fc2 = nn.Linear(reduced, channels)
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        b, c, _, _ = x.shape
        y = self.pool(x).view(b, c)
        y = self.fc1(y)
        y = self.relu(y)
        y = self.fc2(y)
        y = self.sigmoid(y)
        y = y.view(b, c, 1, 1)
        return x * y


class ConvBlock(nn.Module):
    def __init__(self, in_ch, out_ch, use_se=True, pool=True):
        super().__init__()
        self.conv1 = ConvBNReLU(in_ch, out_ch)
        self.conv2 = ConvBNReLU(out_ch, out_ch)
        self.se = SEBlock(out_ch) if use_se else nn.Identity()
        self.pool = nn.MaxPool2d(kernel_size=2, stride=2) if pool else nn.Identity()

    def forward(self, x):
        x = self.conv1(x)
        x = self.conv2(x)
        x = self.se(x)
        x = self.pool(x)
        return x


class CNNEncoder(nn.Module):
    def __init__(self, in_channels=11, use_se_blocks=True):
        super().__init__()
        self.block1 = ConvBlock(in_channels, 32, use_se=use_se_blocks, pool=True)
        self.block2 = ConvBlock(32, 64, use_se=use_se_blocks, pool=True)
        self.block3 = ConvBlock(64, 128, use_se=use_se_blocks, pool=False)
        self.block4 = ConvBlock(128, 256, use_se=use_se_blocks, pool=False)

    def forward(self, x):
        x = self.block1(x)
        x = self.block2(x)
        x = self.block3(x)
        x = self.block4(x)
        return x


class TerrainCNNFusion(nn.Module):
    def __init__(self, in_channels=11, n_handcrafted=15, dropout=0.35,
                 use_se_blocks=True, use_handcrafted=True):
        super().__init__()
        self.encoder = CNNEncoder(in_channels=in_channels, use_se_blocks=use_se_blocks)
        self.dropout = nn.Dropout2d(dropout)
        self.gap = nn.AdaptiveAvgPool2d(1)
        self.use_handcrafted = use_handcrafted

        fusion_dim = 256 + (n_handcrafted if use_handcrafted else 0)

        self.head = nn.Sequential(
            nn.Linear(fusion_dim, 64),
            nn.ReLU(inplace=True),
            nn.Dropout(dropout),
            nn.Linear(64, 1),
        )

    def forward(self, x, feat=None):
        x = self.encoder(x)
        x = self.dropout(x)
        x = self.gap(x).flatten(1)
        if self.use_handcrafted and feat is not None:
            x = torch.cat([x, feat], dim=1)
        return self.head(x).squeeze(1)


def init_weights(module):
    if isinstance(module, nn.Conv2d):
        nn.init.kaiming_normal_(module.weight, mode="fan_out", nonlinearity="relu")
        if module.bias is not None:
            nn.init.zeros_(module.bias)
    elif isinstance(module, nn.Linear):
        nn.init.kaiming_normal_(module.weight, mode="fan_in", nonlinearity="relu")
        if module.bias is not None:
            nn.init.zeros_(module.bias)
    elif isinstance(module, nn.BatchNorm2d):
        nn.init.ones_(module.weight)
        nn.init.zeros_(module.bias)


# ============================================================
# SECTION 8 — TRAIN / EVAL HELPER
# ============================================================

def run_epoch(model, loader, criterion, optimizer=None, scaler=None,
              train=False, use_handcrafted=True):

    model.train() if train else model.eval()

    total_loss = 0.0
    all_probs, all_labels = [], []

    for batch in loader:
        x, hc, soft_labels, hard_labels = batch
        x = x.to(DEVICE, non_blocking=True)
        hc = hc.to(DEVICE, non_blocking=True)
        soft_labels = soft_labels.to(DEVICE, non_blocking=True)
        hard_labels = hard_labels.to(DEVICE, non_blocking=True)

        if train:
            optimizer.zero_grad(set_to_none=True)

        with torch.cuda.amp.autocast(enabled=(DEVICE.type == "cuda")):
            logits = model(x, hc) if use_handcrafted else model(x, None)
            logits = logits.view(-1)
            loss = criterion(logits, soft_labels)

        if train:
            scaler.scale(loss).backward()
            scaler.unscale_(optimizer)
            torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
            scaler.step(optimizer)
            scaler.update()

        total_loss += loss.item() * x.size(0)

        probs = torch.sigmoid(logits)
        all_probs.extend(probs.detach().cpu().numpy().tolist())
        all_labels.extend(hard_labels.detach().cpu().numpy().tolist())

    avg_loss = total_loss / len(loader.dataset)
    all_probs = np.asarray(all_probs, dtype=np.float32)
    all_labels = np.asarray(all_labels, dtype=np.int64)
    preds = (all_probs >= 0.5).astype(int)

    f1 = f1_score(all_labels, preds, zero_division=0)
    try:
        auc = roc_auc_score(all_labels, all_probs)
    except ValueError:
        auc = float("nan")

    return avg_loss, f1, auc, all_probs, all_labels


@torch.no_grad()
def predict_probs(model, loader, use_handcrafted=True):
    model.eval()
    all_probs, all_labels = [], []
    for batch in loader:
        x, hc, _, hard_labels = batch
        x = x.to(DEVICE, non_blocking=True)
        hc = hc.to(DEVICE, non_blocking=True)
        with torch.cuda.amp.autocast(enabled=(DEVICE.type == "cuda")):
            logits = model(x, hc) if use_handcrafted else model(x, None)
            logits = logits.view(-1)
        probs = torch.sigmoid(logits)
        all_probs.extend(probs.detach().cpu().numpy().tolist())
        all_labels.extend(hard_labels.numpy().tolist())
    return np.asarray(all_probs, dtype=np.float32), np.asarray(all_labels, dtype=np.int64)


def best_f1_threshold(probs, labels):
    thresholds = np.linspace(0.05, 0.95, 181)
    best_t, best_f1 = 0.5, -1.0
    for t in thresholds:
        preds = (probs >= t).astype(int)
        f1 = f1_score(labels, preds, zero_division=0)
        if f1 > best_f1:
            best_f1, best_t = f1, t
    return best_t, best_f1


# ============================================================
# SECTION 9 — NESTED, STRATIFIED SUBSETS OF TRAINVAL
# ============================================================

trainval_labels_arr = np.asarray(trainval_labels)
site_idx = np.where(trainval_labels_arr == 1)[0]
nosite_idx = np.where(trainval_labels_arr == 0)[0]

rng = np.random.RandomState(SEED)
rng.shuffle(site_idx)
rng.shuffle(nosite_idx)

print("\n" + "=" * 80)
print("TRAINVAL POOL (train + validation merged)")
print("=" * 80)
print(f"Total samples : {len(trainval_fp)}")
print(f"Site          : {len(site_idx)}")
print(f"No-site       : {len(nosite_idx)}")

subset_indices_by_fraction = {}
for frac in CONFIG["fractions"]:
    n_site = int(round(frac * len(site_idx)))
    n_nosite = int(round(frac * len(nosite_idx)))
    idx = np.concatenate([site_idx[:n_site], nosite_idx[:n_nosite]])
    subset_indices_by_fraction[frac] = idx

fractions_sorted = sorted(CONFIG["fractions"])
for a, b in zip(fractions_sorted[:-1], fractions_sorted[1:]):
    assert set(subset_indices_by_fraction[a]).issubset(
        set(subset_indices_by_fraction[b])
    ), f"Nesting broken between {a} and {b}"

print("\nNesting + stratification check PASSED.")
print("=" * 80)


# ============================================================
# SECTION 10 — FROZEN TEST SET (built once, reused unchanged)
# ============================================================

test_dataset = TerrainPatchDataset(
    filepaths=test_fp,
    hard_labels=test_labels,
    monument_counts=test_monument_counts,
    mean=CHANNEL_MEAN, std=CHANNEL_STD,
    hc_mean=HANDCRAFTED_MEAN, hc_std=HANDCRAFTED_STD,
    augment=False,
    use_handcrafted=CONFIG["use_handcrafted_features"],
    use_channel_dropout=False,
)

test_loader = DataLoader(
    test_dataset, batch_size=CONFIG["batch_size"],
    shuffle=False, num_workers=2, pin_memory=True,
)

print(f"\nFROZEN test set size: {len(test_dataset)} "
      f"(identical across all fractions and all folds, never resampled)\n")


# ============================================================
# SECTION 11 — HELPER: FIND BALANCED SPATIAL 5-FOLD SPLIT
# ============================================================

def find_balanced_spatial_folds(fp, labels, groups, n_folds, n_seeds):
    target_size = len(fp) / n_folds
    target_pos_ratio = np.mean(labels)

    best_score = float("inf")
    best_folds = None
    best_seed = None

    for seed in range(n_seeds):
        sgkf = StratifiedGroupKFold(n_splits=n_folds, shuffle=True, random_state=seed)
        candidate_folds = list(sgkf.split(fp, labels, groups=groups))

        score = 0.0
        for train_idx, val_idx in candidate_folds:
            val_size = len(val_idx)
            val_pos_ratio = np.mean(labels[val_idx])
            size_error = ((val_size - target_size) / target_size) ** 2
            ratio_error = (val_pos_ratio - target_pos_ratio) ** 2
            score += 2.0 * size_error + 5.0 * ratio_error

        if score < best_score:
            best_score = score
            best_folds = candidate_folds
            best_seed = seed

    # ---- hard check: no spatial group ever appears in both
    #      train and val for the same fold ----
    for fold_i, (train_idx, val_idx) in enumerate(best_folds, start=1):
        train_groups = set(groups[train_idx])
        val_groups = set(groups[val_idx])
        overlap = train_groups.intersection(val_groups)
        if overlap:
            raise RuntimeError(
                f"Spatial leakage detected in fold {fold_i}: "
                f"{len(overlap)} overlapping groups."
            )

    return best_folds, best_seed, best_score


# ============================================================
# SECTION 12 — FOR EACH FRACTION: 5-FOLD CV -> ENSEMBLE -> TEST
# ============================================================

fraction_results = []

for frac in CONFIG["fractions"]:

    frac_pct = int(frac * 100)

    idx = subset_indices_by_fraction[frac]
    subset_fp = [trainval_fp[i] for i in idx]
    subset_labels = trainval_labels_arr[idx]
    subset_counts = trainval_monument_counts[idx]
    subset_groups = trainval_groups[idx]

    n_site_sub = int((subset_labels == 1).sum())
    n_nosite_sub = int((subset_labels == 0).sum())

    print("\n" + "#" * 80)
    print(f"# FRACTION = {frac_pct}%   (n={len(subset_fp)}, "
          f"site={n_site_sub}, no_site={n_nosite_sub}, "
          f"site%={100 * n_site_sub / len(subset_fp):.2f}%, "
          f"spatial groups={len(np.unique(subset_groups))})")
    print("#" * 80)

    # --------------------------------------------------------
    # Find a balanced spatial 5-fold split for THIS subset
    # --------------------------------------------------------
    fold_indices, best_seed, best_score = find_balanced_spatial_folds(
        subset_fp, subset_labels, subset_groups,
        n_folds=CONFIG["n_folds"], n_seeds=CONFIG["fold_search_seeds"],
    )
    print(f"Balanced 5-fold split found (seed={best_seed}, score={best_score:.6f})")

    # --------------------------------------------------------
    # Train one model per fold; collect OOF preds + test preds
    # --------------------------------------------------------
    oof_probs = np.zeros(len(subset_fp), dtype=np.float32)
    oof_labels = np.zeros(len(subset_fp), dtype=np.int64)
    fold_test_probs = np.zeros((CONFIG["n_folds"], len(test_fp)), dtype=np.float32)
    fold_val_aucs = []

    for fold_i, (train_idx, val_idx) in enumerate(fold_indices, start=1):

        print(f"\n--- Fraction {frac_pct}% | Fold {fold_i}/{CONFIG['n_folds']} "
              f"(train={len(train_idx)}, val={len(val_idx)}) ---")

        fold_train_fp = [subset_fp[i] for i in train_idx]
        fold_train_labels = subset_labels[train_idx]
        fold_train_counts = subset_counts[train_idx]

        fold_val_fp = [subset_fp[i] for i in val_idx]
        fold_val_labels = subset_labels[val_idx]
        fold_val_counts = subset_counts[val_idx]

        fold_train_dataset = TerrainPatchDataset(
            filepaths=fold_train_fp, hard_labels=fold_train_labels,
            monument_counts=fold_train_counts,
            mean=CHANNEL_MEAN, std=CHANNEL_STD,
            hc_mean=HANDCRAFTED_MEAN, hc_std=HANDCRAFTED_STD,
            augment=True,
            use_handcrafted=CONFIG["use_handcrafted_features"],
            use_channel_dropout=CONFIG["use_channel_dropout"],
            channel_dropout_prob=CONFIG["channel_dropout_prob"],
            channel_dropout_max_channels=CONFIG["channel_dropout_max_channels"],
        )

        fold_val_dataset = TerrainPatchDataset(
            filepaths=fold_val_fp, hard_labels=fold_val_labels,
            monument_counts=fold_val_counts,
            mean=CHANNEL_MEAN, std=CHANNEL_STD,
            hc_mean=HANDCRAFTED_MEAN, hc_std=HANDCRAFTED_STD,
            augment=False,
            use_handcrafted=CONFIG["use_handcrafted_features"],
            use_channel_dropout=False,
        )

        if CONFIG["use_balanced_sampler"]:
            n_pos = int((fold_train_labels == 1).sum())
            n_neg = int((fold_train_labels == 0).sum())
            class_counts = np.array([n_neg, n_pos], dtype=np.float64)
            class_weights = 1.0 / np.maximum(class_counts, 1)
            sample_weights = class_weights[fold_train_labels]
            sampler = WeightedRandomSampler(
                weights=sample_weights, num_samples=len(sample_weights), replacement=True,
            )
            fold_train_loader = DataLoader(
                fold_train_dataset, batch_size=CONFIG["batch_size"],
                sampler=sampler, num_workers=2, pin_memory=True,
                drop_last=(len(fold_train_dataset) > CONFIG["batch_size"]),
            )
        else:
            fold_train_loader = DataLoader(
                fold_train_dataset, batch_size=CONFIG["batch_size"],
                shuffle=True, num_workers=2, pin_memory=True,
                drop_last=(len(fold_train_dataset) > CONFIG["batch_size"]),
            )

        fold_val_loader = DataLoader(
            fold_val_dataset, batch_size=CONFIG["batch_size"],
            shuffle=False, num_workers=2, pin_memory=True,
        )

        model = TerrainCNNFusion(
            in_channels=CONFIG["n_channels"], n_handcrafted=CONFIG["n_handcrafted"],
            dropout=CONFIG["dropout"], use_se_blocks=CONFIG["use_se_blocks"],
            use_handcrafted=CONFIG["use_handcrafted_features"],
        ).to(DEVICE)
        model.apply(init_weights)

        criterion = nn.BCEWithLogitsLoss()
        optimizer = torch.optim.AdamW(
            model.parameters(), lr=CONFIG["learning_rate"], weight_decay=CONFIG["weight_decay"],
        )
        scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
            optimizer, mode="max", factor=CONFIG["scheduler_factor"],
            patience=CONFIG["scheduler_patience"],
        )
        scaler = torch.cuda.amp.GradScaler(enabled=(DEVICE.type == "cuda"))

        best_val_auc = -1.0
        best_state = None
        epochs_no_improve = 0

        for epoch in range(1, CONFIG["epochs"] + 1):

            train_loss, train_f1, train_auc, _, _ = run_epoch(
                model, fold_train_loader, criterion, optimizer=optimizer, scaler=scaler,
                train=True, use_handcrafted=CONFIG["use_handcrafted_features"],
            )
            # early stopping / model selection uses THIS FOLD'S OWN
            # validation set — the frozen test set is never touched here
            val_loss, val_f1, val_auc, _, _ = run_epoch(
                model, fold_val_loader, criterion, optimizer=None, scaler=None,
                train=False, use_handcrafted=CONFIG["use_handcrafted_features"],
            )

            scheduler.step(val_auc)

            print(f"    Epoch {epoch:02d} | train_loss={train_loss:.4f} train_auc={train_auc:.4f} | "
                  f"val_loss={val_loss:.4f} val_auc={val_auc:.4f} val_f1={val_f1:.4f}")

            if val_auc > best_val_auc:
                best_val_auc = val_auc
                best_state = copy.deepcopy(model.state_dict())
                epochs_no_improve = 0
            else:
                epochs_no_improve += 1

            if epochs_no_improve >= CONFIG["early_stopping_patience"]:
                print(f"    Early stopping at epoch {epoch} (fold val AUC, patience={CONFIG['early_stopping_patience']})")
                break

        model.load_state_dict(best_state)

        # ---- save this fold's model ----
        fold_ckpt_path = os.path.join(
            CONFIG["output_dir"], f"frac_{frac_pct}_fold_{fold_i}.pt"
        )
        torch.save(best_state, fold_ckpt_path)

        # ---- OOF predictions for this fold's val set ----
        val_probs, val_labels_arr = predict_probs(
            model, fold_val_loader, use_handcrafted=CONFIG["use_handcrafted_features"]
        )
        oof_probs[val_idx] = val_probs
        oof_labels[val_idx] = val_labels_arr
        fold_val_aucs.append(best_val_auc)

        # ---- this fold's predictions on the FROZEN test set ----
        test_probs_fold, _ = predict_probs(
            model, test_loader, use_handcrafted=CONFIG["use_handcrafted_features"]
        )
        fold_test_probs[fold_i - 1] = test_probs_fold

        print(f"    Fold {fold_i} best val AUC = {best_val_auc:.4f} | model saved to {fold_ckpt_path}")

    # --------------------------------------------------------
    # Lock threshold from OOF predictions ONLY (test untouched)
    # --------------------------------------------------------
    locked_threshold, oof_best_f1 = best_f1_threshold(oof_probs, oof_labels)
    oof_auc = roc_auc_score(oof_labels, oof_probs)

    print(f"\nFraction {frac_pct}% OOF AUC = {oof_auc:.4f} | "
          f"locked threshold = {locked_threshold:.3f} (OOF F1 = {oof_best_f1:.4f})")

    # --------------------------------------------------------
    # Ensemble the 5 fold models on the FROZEN test set
    # --------------------------------------------------------
    ensemble_test_probs = fold_test_probs.mean(axis=0)
    ensemble_test_preds = (ensemble_test_probs >= locked_threshold).astype(int)

    test_auc = roc_auc_score(test_labels, ensemble_test_probs)
    test_accuracy = accuracy_score(test_labels, ensemble_test_preds)
    test_precision = precision_score(test_labels, ensemble_test_preds, zero_division=0)
    test_recall = recall_score(test_labels, ensemble_test_preds, zero_division=0)
    test_f1 = f1_score(test_labels, ensemble_test_preds, zero_division=0)

    cm = confusion_matrix(test_labels, ensemble_test_preds)
    cm_path = os.path.join(CONFIG["output_dir"], f"confusion_matrix_frac_{frac_pct}.csv")
    pd.DataFrame(
        cm, index=["true_no_site", "true_site"], columns=["pred_no_site", "pred_site"]
    ).to_csv(cm_path)

    oof_df = pd.DataFrame({
        "patch": [os.path.basename(p) for p in subset_fp],
        "oof_prob": oof_probs,
        "oof_label": oof_labels,
    })
    oof_path = os.path.join(CONFIG["output_dir"], f"oof_predictions_frac_{frac_pct}.csv")
    oof_df.to_csv(oof_path, index=False)

    fraction_results.append({
        "fraction": frac,
        "n_train_samples": len(subset_fp),
        "n_site": n_site_sub,
        "n_no_site": n_nosite_sub,
        "n_spatial_groups": len(np.unique(subset_groups)),
        "oof_auc": oof_auc,
        "locked_threshold": locked_threshold,
        "mean_fold_val_auc": float(np.mean(fold_val_aucs)),
        "test_auc": test_auc,
        "test_accuracy": test_accuracy,
        "test_precision": test_precision,
        "test_recall": test_recall,
        "test_f1": test_f1,
        "confusion_matrix_path": cm_path,
        "oof_predictions_path": oof_path,
    })

    print()
    print("-" * 80)
    print(f"FRACTION {frac_pct}% — ENSEMBLE (5-fold) FINAL RESULTS ON FROZEN TEST SET")
    print("-" * 80)
    print(f"  Accuracy  : {test_accuracy:.4f}  ({test_accuracy * 100:.2f}%)")
    print(f"  Precision : {test_precision:.4f}  ({test_precision * 100:.2f}%)")
    print(f"  Recall    : {test_recall:.4f}  ({test_recall * 100:.2f}%)")
    print(f"  F1 Score  : {test_f1:.4f}  ({test_f1 * 100:.2f}%)")
    print(f"  ROC-AUC   : {test_auc:.4f}  ({test_auc * 100:.2f}%)")
    print(f"  (OOF AUC across folds: {oof_auc:.4f}, threshold locked at {locked_threshold:.3f})")
    print("-" * 80 + "\n")


# ============================================================
# SECTION 13 — SAVE FINAL COMPARISON RESULTS
# ============================================================

results_df = pd.DataFrame(fraction_results)

print("=" * 80)
print("LEARNING CURVE RESULTS — 5-FOLD ENSEMBLE, FROZEN TEST SET")
print("=" * 80)

for r in fraction_results:
    print(
        f"{int(r['fraction']*100):3d}% train (n={r['n_train_samples']:5d}) | "
        f"Accuracy={r['test_accuracy']:.4f} | "
        f"Precision={r['test_precision']:.4f} | "
        f"Recall={r['test_recall']:.4f} | "
        f"F1={r['test_f1']:.4f} | "
        f"AUC={r['test_auc']:.4f} | "
        f"OOF_AUC={r['oof_auc']:.4f}"
    )

print()
print(results_df.to_string(index=False))

results_csv_path = os.path.join(CONFIG["output_dir"], "learning_curve_5fold_results.csv")
results_df.to_csv(results_csv_path, index=False)
print(f"\nResults table saved to: {results_csv_path}")


# ============================================================
# SECTION 14 — SAVE COMPARISON PLOTS
# ============================================================

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

axes[0].plot(results_df["fraction"] * 100, results_df["test_auc"], marker="o", label="Test AUC (ensemble)")
axes[0].plot(results_df["fraction"] * 100, results_df["oof_auc"], marker="d", linestyle="--", label="OOF AUC")
axes[0].plot(results_df["fraction"] * 100, results_df["test_f1"], marker="s", label="Test F1")
axes[0].set_xlabel("Training data used (%)")
axes[0].set_ylabel("Score")
axes[0].set_title("AUC / F1 vs Training Data Fraction (5-fold ensemble, frozen test set)")
axes[0].legend()
axes[0].grid(alpha=0.3)

axes[1].plot(results_df["fraction"] * 100, results_df["test_accuracy"], marker="o", label="Accuracy")
axes[1].plot(results_df["fraction"] * 100, results_df["test_precision"], marker="s", label="Precision")
axes[1].plot(results_df["fraction"] * 100, results_df["test_recall"], marker="^", label="Recall")
axes[1].set_xlabel("Training data used (%)")
axes[1].set_ylabel("Score")
axes[1].set_title("Test Accuracy / Precision / Recall vs Training Data Fraction")
axes[1].legend()
axes[1].grid(alpha=0.3)

plt.tight_layout()

plot_path = os.path.join(CONFIG["output_dir"], "learning_curve_5fold_plots.png")
plt.savefig(plot_path, dpi=150, bbox_inches="tight")
plt.show()

print(f"Plots saved to: {plot_path}")

print("\n" + "=" * 80)
print("ALL OUTPUTS SAVED UNDER:", CONFIG["output_dir"])
print("=" * 80)
print(" - learning_curve_5fold_results.csv    (metrics table, all fractions)")
print(" - confusion_matrix_frac_XX.csv        (per fraction, ensemble on test)")
print(" - oof_predictions_frac_XX.csv         (per fraction, all 5 folds' OOF preds)")
print(" - frac_XX_fold_Y.pt                   (best model weights, per fraction per fold)")
print(" - learning_curve_5fold_plots.png      (comparison charts)")
print("=" * 80)
print("\nLEARNING CURVE + 5-FOLD CV EXPERIMENT COMPLETE.")
```

```python
# ============================================================
# EXTRA CELL — VISUALIZE 30% / 50% / 70% SPLITS AGAINST EACH OTHER
# ============================================================
# Run this AFTER the main learning-curve script.
# It first tries to reuse `results_df` already in memory; if that
# isn't available (e.g. fresh kernel), it reloads everything from
# the saved CSV/confusion-matrix files instead.
# ============================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

OUTPUT_DIR = CONFIG["output_dir"] if "CONFIG" in globals() else \
    "/kaggle/working/learning_curve_5fold_results"

# ------------------------------------------------------------
# Load results (reuse in-memory results_df if present, else reload)
# ------------------------------------------------------------

if "results_df" in globals():
    viz_df = results_df.copy()
else:
    viz_df = pd.read_csv(
        os.path.join(OUTPUT_DIR, "learning_curve_5fold_results.csv")
    )

viz_df = viz_df.sort_values("fraction").reset_index(drop=True)
viz_df["pct_label"] = (viz_df["fraction"] * 100).astype(int).astype(str) + "%"

print(viz_df[[
    "pct_label", "n_train_samples", "test_accuracy",
    "test_precision", "test_recall", "test_f1", "test_auc", "oof_auc",
]].to_string(index=False))


# ============================================================
# PLOT 1 — Grouped bar chart: all 5 metrics side by side per split
# ============================================================

metrics = ["test_accuracy", "test_precision", "test_recall", "test_f1", "test_auc"]
metric_labels = ["Accuracy", "Precision", "Recall", "F1", "AUC"]

x = np.arange(len(metrics))
width = 0.25
colors = ["#4C72B0", "#DD8452", "#55A868"]

fig, ax = plt.subplots(figsize=(11, 6))

for i, (_, row) in enumerate(viz_df.iterrows()):
    offset = (i - (len(viz_df) - 1) / 2) * width
    values = [row[m] for m in metrics]
    bars = ax.bar(x + offset, values, width, label=row["pct_label"], color=colors[i % len(colors)])
    for b, v in zip(bars, values):
        ax.text(b.get_x() + b.get_width() / 2, v + 0.01, f"{v:.3f}",
                ha="center", va="bottom", fontsize=8)

ax.set_xticks(x)
ax.set_xticklabels(metric_labels)
ax.set_ylim(0, 1.05)
ax.set_ylabel("Score")
ax.set_title("Test Metrics Comparison Across Training Data Splits (frozen test set)")
ax.legend(title="Training data used")
ax.grid(axis="y", alpha=0.3)
plt.tight_layout()

bar_path = os.path.join(OUTPUT_DIR, "comparison_grouped_bar.png")
plt.savefig(bar_path, dpi=150, bbox_inches="tight")
plt.show()
print(f"Saved: {bar_path}")


# ============================================================
# PLOT 2 — Line chart: metric trend as training data increases
# ============================================================

fig, ax = plt.subplots(figsize=(9, 6))

markers = ["o", "s", "^", "D", "P"]
for m, lbl, mk in zip(metrics, metric_labels, markers):
    ax.plot(viz_df["fraction"] * 100, viz_df[m], marker=mk, label=lbl, linewidth=2)

ax.set_xlabel("Training data used (%)")
ax.set_ylabel("Score")
ax.set_title("Metric Trends as Training Data Increases")
ax.set_xticks(viz_df["fraction"] * 100)
ax.legend()
ax.grid(alpha=0.3)
plt.tight_layout()

line_path = os.path.join(OUTPUT_DIR, "comparison_line_trend.png")
plt.savefig(line_path, dpi=150, bbox_inches="tight")
plt.show()
print(f"Saved: {line_path}")


# ============================================================
# PLOT 3 — OOF AUC vs Test AUC per split (generalization gap)
# ============================================================

fig, ax = plt.subplots(figsize=(8, 5.5))

x_pos = np.arange(len(viz_df))
w = 0.35

ax.bar(x_pos - w / 2, viz_df["oof_auc"], w, label="OOF AUC (5-fold)", color="#8172B2")
ax.bar(x_pos + w / 2, viz_df["test_auc"], w, label="Test AUC (ensemble)", color="#C44E52")

for i, (oof, test) in enumerate(zip(viz_df["oof_auc"], viz_df["test_auc"])):
    ax.text(i - w / 2, oof + 0.01, f"{oof:.3f}", ha="center", fontsize=8)
    ax.text(i + w / 2, test + 0.01, f"{test:.3f}", ha="center", fontsize=8)

ax.set_xticks(x_pos)
ax.set_xticklabels(viz_df["pct_label"])
ax.set_ylim(0, 1.05)
ax.set_xlabel("Training data used")
ax.set_ylabel("AUC")
ax.set_title("OOF AUC vs Frozen-Test AUC per Split (generalization check)")
ax.legend()
ax.grid(axis="y", alpha=0.3)
plt.tight_layout()

gap_path = os.path.join(OUTPUT_DIR, "comparison_oof_vs_test_auc.png")
plt.savefig(gap_path, dpi=150, bbox_inches="tight")
plt.show()
print(f"Saved: {gap_path}")


# ============================================================
# PLOT 4 — Confusion matrices side by side, one per split
# ============================================================

fig, axes = plt.subplots(1, len(viz_df), figsize=(5 * len(viz_df), 4.5))
if len(viz_df) == 1:
    axes = [axes]

for ax, (_, row) in zip(axes, viz_df.iterrows()):
    cm_path = row["confusion_matrix_path"] if "confusion_matrix_path" in row else \
        os.path.join(OUTPUT_DIR, f"confusion_matrix_frac_{int(row['fraction']*100)}.csv")
    cm_df = pd.read_csv(cm_path, index_col=0)
    cm = cm_df.values

    im = ax.imshow(cm, cmap="Blues")
    ax.set_title(f"{row['pct_label']} training data")
    ax.set_xticks([0, 1])
    ax.set_yticks([0, 1])
    ax.set_xticklabels(["Pred: no_site", "Pred: site"])
    ax.set_yticklabels(["True: no_site", "True: site"])

    for i in range(cm.shape[0]):
        for j in range(cm.shape[1]):
            ax.text(j, i, str(cm[i, j]), ha="center", va="center",
                     color="white" if cm[i, j] > cm.max() / 2 else "black",
                     fontsize=13, fontweight="bold")

plt.tight_layout()

cm_grid_path = os.path.join(OUTPUT_DIR, "comparison_confusion_matrices.png")
plt.savefig(cm_grid_path, dpi=150, bbox_inches="tight")
plt.show()
print(f"Saved: {cm_grid_path}")


# ============================================================
# PLOT 5 — Training set size vs training data fraction (context)
# ============================================================

fig, ax = plt.subplots(figsize=(7, 5))
bars = ax.bar(viz_df["pct_label"], viz_df["n_train_samples"], color="#64B5CD")
for b, v in zip(bars, viz_df["n_train_samples"]):
    ax.text(b.get_x() + b.get_width() / 2, v + 1, str(int(v)), ha="center", fontsize=9)
ax.set_ylabel("Number of training samples")
ax.set_title("Actual Training Set Size per Split (nested, stratified)")
ax.grid(axis="y", alpha=0.3)
plt.tight_layout()

size_path = os.path.join(OUTPUT_DIR, "comparison_train_size.png")
plt.savefig(size_path, dpi=150, bbox_inches="tight")
plt.show()
print(f"Saved: {size_path}")

print("\nAll comparison visuals saved under:", OUTPUT_DIR)
```
